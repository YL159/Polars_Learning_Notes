### 2.4.1 Temporary Configuration Using a Context Manager

```python
with pl.Config(verbose=True):
    # Polars operation you want to see the verbose logging of
    pass
```

```python
import random
import string

# generate a random string of size 'length'
def generate_random_string(length: int) -> str:
    return "".join(random.choice(string.ascii_letters) for i in range(length))


data = {}
# generate 10 columns of 5 lines of random string of length 50
for i in range(1, 11):
    data[f"column_{i}"] = [generate_random_string(50) for _ in range(5)]  

df = pl.DataFrame(data)
```

```python
df
```

```python
# temporal context: table_cols=-1 show all cols
# format_str_lengths=4 show string head 4 chars
with pl.Config(tbl_cols=-1, fmt_str_lengths=4):
    print(df)
```

```python
# customized Context Manager class, realize __enter__, __exit__ methods, to use 'with'
class YourContextManager:
    def __enter__(self):
        print("Entering context")

    def __exit__(self, type, value, traceback):
        print("Exiting context")


with YourContextManager():
    print("Your code")
```
